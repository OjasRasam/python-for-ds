## Iterators
### An iterator is like a bookmark in a book that remembers where you left off.

```python
"""What makes an iterator?
    Shoul implement - __iter__() , __next__()"""

# Lets make an iterator to understand

class MyIterator:
    def __init__(self):
        self.n = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.n <= 3:
            value = self.n
            self.n += 1
            return value
        else:
            raise StopIteration

# Test
for i in MyIterator():
    print(i)

```
