## Generators
### A generator is an iterator you write with a function instead of a class
Use yield instead of return

```python
# Example
def my_gen():
    yield 1
    yield 2
    yield 3

# Test
for x in my_gen():
    print(x)
```

```python
def infinite_numbers():
    n = 1
    while True:
        yield n
        n += 1

```
