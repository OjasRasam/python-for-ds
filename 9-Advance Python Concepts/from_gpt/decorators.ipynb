## Decorators
### A decorator is like wrappping a gift
* The gift = your original function
* The Wrapping = extra features (logging, timing, caching)
* You don't modify the gift
* You wrap it

```python
# Basic Decorator
def my_decorator(func):
    def wrapper():
        print("Before Function")
        func()
        print("After Function")
    return wrapper
```

```python
# Use it:
@my_decorator
def greet():
    print("Hello")

# Calling
greet()
```
